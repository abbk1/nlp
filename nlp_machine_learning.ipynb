# Text Classification

## Using import and naive bias to clasify text

```python
# Training model
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
```

```python
from pathlib import Path
# Find the folder where your notebook is located
NOTEBOOK_DIR = Path.cwd()
# If using pathlib, construct the path dynamically
# If the "Data" folder is sitting inside "Course Materials" relative to the notebook:
# file_path = NOTEBOOK_DIR / "Course Materials" / "Data" / "Popchip.xlsx"
data_path = NOTEBOOK_DIR / "Course Materials" / "Data" / "Popchip_Reviews.xlsx"
```

```python
data_df = pd.read_excel(data_path)
data_df.head(2)
```

```python
data_df.Priority.value_counts()
```

```python
## importing custom functions for text preprocessing
import text_preprocessing as tp
```

```python
data_df.head(2)
```

```python
data_df['Clean_Text'] = tp.lower_replace_special_char(data_df.Text)
```

```python
data_df.head(2)
```

```python
data_df['Clean_Text'] = data_df.Clean_Text.apply(tp.token_lemma_stopw_join)
```

```python
data_df.head(2)
```

```python
vectorizer = CountVectorizer(stop_words="english", ngram_range=(1,2), min_df=.2)
X = vectorizer.fit_transform(data_df.Clean_Text)
X_df = pd.DataFrame(X.toarray(),columns=vectorizer.get_feature_names_out())
X_df
```

```python
# X_df = data_df.Clean_Text
y = data_df.Priority
y.head()
```

```python
# ---------------------------------------------------
# Train-Test Split
# ---------------------------------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X_df,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
X_train.shape
```

```python
# ---------------------------------------------------
# Train
# ---------------------------------------------------
nb_model = MultinomialNB()
nb_model.fit(X_train, y_train)


# ---------------------------------------------------
# TF-IDF + Naive Bayes Pipeline
# ---------------------------------------------------
# nb_model = Pipeline([
#     ("vectorizer", CountVectorizer(
#         lowercase=True,
#         stop_words="english",
#         ngram_range=(1,2),
#         min_df=2
#     )),
#     ("classifier", MultinomialNB())
# ])
```

```python
# ---------------------------------------------------
# Prediction
# ---------------------------------------------------
y_pred = nb_model.predict(X_test)
```

```python
# ---------------------------------------------------
# Evaluation
# ---------------------------------------------------
print("Accuracy:", accuracy_score(y_test, y_pred))

print("\nClassification Report")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix")
print(confusion_matrix(y_test, y_pred))
```

# NMF Topic Modelling

```python
data_df.head(2)
```

```python
tfidf = TfidfVectorizer(stop_words="english", min_df=0.05, max_df=.2)
Xt2 = tfidf.fit_transform(data_df.Clean_Text)
Xt2_df = pd.DataFrame(Xt2.toarray(),columns=tfidf.get_feature_names_out())
Xt2_df
```

```python
data_df.shape
```

```python
from sklearn.decomposition import NMF

nmf = NMF(n_components=2)
W = nmf.fit_transform(Xt2_df) # Documnets-Topics
H = nmf.components_ # Topics-Terms
```

```python
H.shape
```

```python
# function to display the top terms for each topic
def display_topics(H, num_words=10):
    for topic_num, topic_array in enumerate(H):
        top_features = topic_array.argsort()[::-1][:num_words]
        top_words = [tfidf.get_feature_names_out()[i] for i in top_features]
        print("Topic", topic_num+1, ":", ', '.join(top_words))
```

```python
display_topics(H)
```
