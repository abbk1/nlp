```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
analyzer = SentimentIntensityAnalyzer()
sentiment = analyzer.polarity_scores(phrase)
sentiment
```

```python
# define a function to classify the sentiment based on the compound score
def sentiment_score(text):
    analyzer = SentimentIntensityAnalyzer()
    return analyzer.polarity_scores(text)['compound']
    
```

```python
df['sentiment_score'] = df['movie_info'].apply(sentiment_score)
```

# Text Clasification

1. Clean and normalize the _movie_info_ column using the _maven_text_preprocessing.py_ module
2. Create a Count Vectorizer
* Remove stop words
* Set the minimum document frequency to 10%
3. Create a Naïve Bayes model and a Logistic Regression model to predict which movies are directed by women vs men using the CV
4. Compare their accuracy scores and classification reports
5. Using the better performing model, return the top 5 movies that the model predicts are most likely directed by a women

```python
# Training model
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
```

```python
df['director_gender'].value_counts()
# df['director_gender'].value_counts(normalize=True)*100 in percentage
```

```python
import text_preprocessing as tp
```

```python
df['clean_movie_info'] = tp.nlp_pipeline(df['movie_info'])
```

```python
df[['movie_info','clean_movie_info']].head(2)
```

```python
# Count Vectorizer
vectorizer = CountVectorizer(stop_words="english", min_df=.10)
X = vectorizer.fit_transform(df.clean_movie_info)
X_df = pd.DataFrame(X.toarray(),columns=vectorizer.get_feature_names_out())


# TF-IDF Vectorizer
tfidf = TfidfVectorizer(stop_words="english", min_df=.10)
Xt = tfidf.fit_transform(df.clean_movie_info)
Xt_df = pd.DataFrame(Xt.toarray(),columns=tfidf.get_feature_names_out())
```

```python
#CountVectorizer
X_df.head()
```

```python
#TF-IDF Vectorizer
Xt_df.head()
```

```python
y = df.director_gender ## use with CountVectorizer
yt = df.director_gender ## Use with TF-IDF Vectorizer
y.head(2)
```

```python
# ---------------------------------------------------
# Count Vectorizer Train-Test Split
# ---------------------------------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X_df,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
# ---------------------------------------------------
# TF-IDF Vectorizer Train-Test Split
# ---------------------------------------------------
Xt_train, Xt_test, yt_train, yt_test = train_test_split(
    Xt_df,
    yt,
    test_size=0.2,
    random_state=42,
    stratify=y
)
X_train.shape, Xt_train.shape
```

# Model Comparison for Count Vectorizer

```python
# ---------------------------------------------------
# Train
# ---------------------------------------------------
nb_model = MultinomialNB()
nb_model.fit(X_train, y_train)
```

```python
# ---------------------------------------------------
# Train
# ---------------------------------------------------
model_lr = LogisticRegression()
model_lr.fit(X_train, y_train)
```

```python
# ---------------------------------------------------
# Prediction
# ---------------------------------------------------
y_pred_nb = nb_model.predict(X_test)
# ---------------------------------------------------
# Prediction
# ---------------------------------------------------
y_pred_lr = model_lr.predict(X_test)
```

```python
# ---------------------------------------------------
# Evaluation NB
# ---------------------------------------------------
print("\nClassification Report")
print(classification_report(y_test, y_pred_nb))

print("\nConfusion Matrix")
print(confusion_matrix(y_test, y_pred_nb))

print("Accuracy:", accuracy_score(y_test, y_pred_nb),"\n")

print("<<<<<<<<<<<<<<<<<<<<<<<<<<<<Below Is for LOgistic Regression>>>>>>>>>>>>>>>>>>>>>>>>>>>>")
# ---------------------------------------------------
# Evaluation LR
# ---------------------------------------------------
print("\nClassification Report LR")
print(classification_report(y_test, y_pred_lr))

print("\nConfusion Matrix LR")
print(confusion_matrix(y_test, y_pred_lr))

print("Accuracy LR:", accuracy_score(y_test, y_pred_lr))
```

# Topic Modelling Assignment

1. Using the same preprocessed data as the last assignment, create a Tfidf Vectorizer
* Remove stop words
* Start with min_df = 0.05 and max_df=0.2
2. Create an NMF model to find the main topics in the movie descriptions
* Start with n_components=2
3. Tweak the model by updating the Tfidf Vectorizer parameters and number of topics
4. Interpret and name the topics
5. For two of the topics, return the top movies that contain the topic

```python
# TF-IDF Vectorizer
tfidf1 = TfidfVectorizer(stop_words="english", min_df=0.05, max_df=0.2)
Xt1 = tfidf1.fit_transform(df.clean_movie_info)
Xt_df1 = pd.DataFrame(Xt1.toarray(),columns=tfidf1.get_feature_names_out())
```

```python
Xt_df1
```

```python
Xt_df1.shape
```

```python
from sklearn.decomposition import NMF

nmf = NMF(n_components=5, max_iter=500, random_state=42)
W = nmf.fit_transform(Xt_df1) # Documnets-Topics
H = nmf.components_ # Topics-Terms
```

```python
H.shape
```

```python
# function to display the top terms for each topic
def display_topics(H, num_words=10):
    for topic_num, topic_array in enumerate(H):
        top_features = topic_array.argsort()[::-1][:num_words]
        top_words = [tfidf1.get_feature_names_out()[i] for i in top_features]
        print("Topic", topic_num+1, ":", ', '.join(top_words))
```

```python
display_topics(H)
```

```python
topics = pd.DataFrame(W, columns=['Growth','Entertaiment','Action','Dramatic-Fiction','Family'])
topics.head()
```

```python
print(topics.iloc[:, 1:].dtypes)
```

```python
doc_topics = pd.concat([df_sort, topics], axis=1)
doc_topics.head(1)
```

```python
# in each row of topics that has highest value assigned its label to new Topic column as
doc_topics['Topics'] = doc_topics.iloc[:,5:].idxmax(axis=1)
doc_topics.head(1)
```
